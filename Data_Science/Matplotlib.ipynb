```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
num = np.linspace(0,10,11)
sq = num**2
```

```python
plt.plot(num,sq)#(x=x,y=y)
plt.title("Square Graph")
plt.xlabel("Number")
plt.ylabel("Square")
```

Sub-Plots

```python
plt.subplot(2,2,1)
plt.plot(num,sq)
plt.subplot(2,2,2)
plt.plot(num*num,sq)
plt.subplot(2,2,3)
plt.plot(sq,num)
plt.subplot(2,2,4)
plt.plot(sq,num)
#
```

Object Oriented Way

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(num,sq)
axis2 = fig.add_axes([0.3,0.1,0.3,0.3])
axis2.plot(num,sq)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(num,sq)
axis1.set_xlabel("Number")
axis1.set_ylabel("Square")
plt.savefig('basicplot.png')
```

Distribution Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df['size'].unique()
```

```python
sns.distplot(df['total_bill'],bins = 10, kde = True) #def: 20
```

```python
plt.subplot(1,2,1)
sns.displot(df['total_bill'])
plt.subplot(1,2,2)
sns.histplot(df['tip'])
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'scatter')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'kde') #hex,hist,reg
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='sex')
```

```python
sns.pairplot(df, hue='size', palette='rainbow')
```

Categorical

```python
sns.countplot(df['sex'])
```

```python
sns.countplot(x=df['sex'])
```

```python
sns.countplot(x=df['sex'], hue=df['smoker'])
```

Barplot: Two Args: y=quantitavive x=categorical

```python
sns.barplot(x='sex',y='total_bill',data=df) #def Estimator:Mean
```

```python

sns.barplot(x='sex',y='total_bill',data=df, estimator=np.median) #def Estimator:Mean
```
