Five-Number Summary

```python
import numpy as np
arr = np.array([0,2,3,4,5,6,7,8,9,12,13,14,15,16,18,23,24,25,28,29,34,88])
arr
```

```python
q1 = np.percentile(arr, 25)
q2 = np.percentile(arr, 50)
q3 = np.percentile(arr, 75)
iqr = q3-q1
uf = q3+1.5*iqr
lf = q1-1.5*iqr

arr = arr[(arr<=uf) & (arr>=lf)]
arr
```

```python
import seaborn as sns
```

```python
sns.boxplot(x = arr)
```

Z-Test

```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174,168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))
z_score
```

```python
p_value = 2 * ( 1 - norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```

T-Test

```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174,168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)
n = len(sample)
dof = n-1
```

```python
t_stats = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
t_stats
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_stats), dof))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```
