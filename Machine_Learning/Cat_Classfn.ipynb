```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
x = df.drop('species', axis=1)
y = df['species']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train, y_train)
```

```python
knn.score(x_test, y_test)
```

```python
from sklearn.svm import SVC
```

```python
svm = SVC(gamma = 'auto') #1.0 for C=10
svm.fit(x_train, y_train)
```

```python
svm.score(x_test, y_test)
```

GRID SEARCH CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((svm),{
    'C':[1,10,20,30],
    'kernel':['rbf', 'linear']
}, cv=5, return_train_score=False)
```

```python
classifier.fit(x_train, y_train)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

RANDOM SEARCH CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
random_classifier = RandomizedSearchCV((svm),{
    'C':[1,10,20,30],
    'kernel':['rbf', 'linear']
}, n_iter = 4, cv=5, return_train_score=False)
```

```python
random_classifier.fit(x_train, y_train)
```

```python
random_classifier.cv_results_
```

```python
pd.DataFrame(random_classifier.cv_results_)[['param_C','param_kernel','mean_test_score']]
```
