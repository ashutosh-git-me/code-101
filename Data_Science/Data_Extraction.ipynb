```python
#Max Population
df[df['population']==df['population'].max()]['country']
```

```python
#Top 5 Highest Democratic Score
df.sort_values(by='democracy_score',ascending=False).head()['country']
```

```python
df.columns
```

```python
#Leader of II most populous
df[df['population']==df['population'].nlargest(2).iloc[1]]['political_leader']
```

```python
#unknown political leaders
df[df['political_leader'].isna()]['country']
```

```python
df[df['political_leader'].isna()]['country'].count()
```

```python
#Republic in Full name(country_long)
count = 0
def rep(txt):
  global count
  if 'republic' in txt.lower():
    count+=1
  return txt
df['country_long'].apply(rep)
print(count)
```
