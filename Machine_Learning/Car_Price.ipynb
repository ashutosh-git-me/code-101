```python
x = df.drop(columns=['price'],axis =1)
y = df['price']
```

```python
df.columns
```

```python
x_one_encode = pd.get_dummies(x,columns=['model','transmission','fuelType'] ,drop_first=True)
x_one_encode = x_one_encode.astype(int)
x_one_encode
```

```python
from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()
columns = ['model','transmission','fuelType']
x_label = x
for col in columns:
    x_label[col] = encode.fit_transform(x_label[col])
x_label
```

```python
x_label['fuelType'].value_counts()
```

```python
x_label['model'].value_counts()
```

Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
num_cols = ['year','mileage','tax','mpg','engineSize']
x_one_encode[num_cols] = scaler.fit_transform(x_one_encode[num_cols])
x_one_encode
```

```python
x_label = pd.DataFrame(
    scaler.fit_transform(x_label.values),
    columns=x_label.columns,
    index=x_label.index,
)
x_label
```

Model 1: x_one_encode

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

```

```python
x_train,x_test,y_train,y_test = train_test_split(x_one_encode,y,test_size=0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test,y_pred)
r2
```

```python
n= x_train.shape[0]
p = x_train.shape[1]
adjusted_r2 = 1-(1-r2)*(n-1)/(n-p-1)
adjusted_r2
```

Model 2: x_label

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_label,y,test_size=0.2,random_state=42)
```

```python
model2 = LinearRegression()
model2.fit(x_train,y_train)
```

```python
y_pred = model2.predict(x_test)
y_pred
```
