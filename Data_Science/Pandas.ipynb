Merge & Concat

```python
import numpy as np
import pandas as pd
```

```python
employee = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Jane', 'Bob', 'Alice', 'Mike'],
    'department_name': ['HR', 'IT', 'Finance', 'IT', 'Sales']
})
salary = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [50000, 60000, 55000, 70000, 65000],
    'bonus': [1000, 2000, 1500, 2500, 3000]
})
```

```python
employee
```

```python
salary
```

```python
#pd.merge(employee,salary) --->default
pd.merge(employee,salary,on="employee_id")
```

```python
pd.merge(employee,salary,on="employee_id",how="outer")
```

```python
pd.merge(employee,salary,on="employee_id",how="left")
```

GroupBy

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S2', 'S1', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 300, 250, 350, 500, 200],
    'Quantity': [10, 20, 15, 30, 25, 35, 50, 20],
    'Date': pd.date_range(start='2026-07-26', periods=8)
}
df = pd.DataFrame(data)
df
```

```python
cat = df.groupby('Category')
for i,j in cat:
  print(i)
  print(j)
```

```python
cat = df.groupby('Category')['Sales'].sum()
print(cat)
```

```python
cat = df.groupby(['Category','Store'])['Sales'].sum()
print(cat)
```

```python
df['Sales'].agg(['sum','mean','max','min'])
```

Pivot Tables

```python
import numpy as np
import pandas as pd
import random
```

```python
data = {
    'Date': pd.date_range(start='2026-06-26', periods=20),
    'Product': ['A','B','C','D'] * 5,
    'Region': ['East','West','North','North','South','West','East','South','West','East','North','West','East','South','West','East','South','West','East','South'],
    'Sales': np.random.randint(100,1000,20),
    'Units': np.random.randint(10,100,20),
    'Rep': ['John','Jane','Bob','Alice','Mike','Jane','Mike','Bob','Alice','John','Bob','Alice','John','Jane','Mike','Bob','Alice','John','Jane','Mike']

}
df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='sum')#aggfunc=mean by default
```

```python
pivot2 = pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product',aggfunc='sum')
pivot2
```

```python
pd.crosstab(df['Region'],df['Product'])
```
